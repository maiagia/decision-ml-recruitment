```python
import json
import pandas as pd

# Carregar o JSON
with open("../data/vagas.json", encoding="utf-8") as f:
    vagas = json.load(f)

# Se for um dicionário com nome da vaga como chave:
if isinstance(vagas, dict):
    # Transforma em DataFrame
    df_vagas = pd.DataFrame.from_dict(vagas, orient="index")
    df_vagas.index.name = "nome_vaga"
    df_vagas.reset_index(inplace=True)
else:
    # Caso seja uma lista de dicionários
    df_vagas = pd.DataFrame(vagas)

# Visualiza as 10 primeiras vagas
df_vagas.head(10)
```
