```python
import json
import pandas as pd

# Carrega o arquivo JSON
with open("../data/prospects.json", encoding="utf-8") as f:
    prospects_data = json.load(f)

# Lista para coletar as situações reais
situacoes = []

# Percorre todas as vagas e seus prospects
for vaga_id, vaga_data in prospects_data.items():
    for prospect in vaga_data.get("prospects", []):
        # Corrigido: chave correta com 'dado' no final
        situacao = prospect.get("situacao_candidado")
        if situacao:
            situacoes.append(situacao.strip())

# Cria DataFrame com situações únicas
df_situacoes = pd.DataFrame(sorted(set(situacoes)), columns=["situacao_candidado"])

# Exibe os resultados
print(f"🔍 Total de situações únicas: {len(df_situacoes)}")
print(df_situacoes.head(30))

# (Opcional) exporta para análise externa
df_situacoes.to_csv("../output/situacoes_unicas.csv", index=False, encoding="utf-8-sig")
print("✅ Situações exportadas para: ../output/situacoes_unicas.csv")
```
